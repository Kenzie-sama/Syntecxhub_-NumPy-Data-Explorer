<h1>Importing the Numpy <br>Loading and printing the dataset</h1>


```python

import numpy as np

marks = np.load('../data/student_marks.npy')

'''
Subject Indexing:

0 -> Math
1 -> Science
2 -> English
3 -> Computer
'''

print("Student Marks:")
print(marks)
```

<h2>Basic Array Exploration</h2>

```python
print('Shape: ', marks.shape)
print('Dimensions: ', marks.ndim)
print('Size: ', marks.size)
print('Data Type: ', marks.dtype)
```

<h2>Indexing and Slicing</h2>

```python
print('First Student Marks: ', marks[0])
print('First Student Math Marks: ', marks[0,0])
print('Third Student Computer Marks: ', marks[2,3])
```

```python
print('Marks of first three students: \n', marks[:3])
print('Only Math marks of students: \n', marks[:,0])
print('Math and Science marks of students: \n', marks[:,0:2])
print('Marks of last two students: \n', marks[-2:])
```

<h2>Mathematical Operation</h2>

```python
bonus_marks = marks + 5

print('Increased Marks: \n', bonus_marks)
```

```python
print(marks * 2)
```

<h2>Axis-wise Operation</h2>
<p> Axis = 0 --> down the rows --> column-wise <br> Axis = 1 --> across the column --> row-wise</p>

```python
subject_avg = np.mean(marks, axis = 0)
student_avg = np.mean(marks, axis = 1)

print('Avg Marks by Subject: ', subject_avg)
print('Avg Marks by Student: ', student_avg)
```

<h2>Statistical Analysis</h2>

```python
print('Mean: ', np.mean(marks))
print('Median: ', np.median(marks))
print('Maximum: ', np.max(marks))
print('Minimum: ', np.min(marks))
print('Standard Deviation ', np.std(marks))
print('Total: ', np.sum(marks))
```

<h4>Subject-wise</h4>

```python
print('Subject Mean: ', np.mean(marks, axis= 0))
print('Subject Median: ', np.median(marks, axis= 0))
print('Subject Maximum: ', np.max(marks, axis= 0))
print('Subject Minimum: ', np.min(marks, axis= 0))
```

<h2>Broadcasting</h2>

```python
#Adding different bonus marks in each subject

bonus = np.array([5, 4, 6, 2])

new_marks = marks + bonus

print(new_marks)
```

<h2>Saving and loading the dataset</h2>

```python
np.save('../data/updated_student_marks.npy', new_marks)
```

```python
loaded_marks = np.load('../data/updated_student_marks.npy')
print(loaded_marks)
```

<h2>Numpy VS Python List</h2>

```python
import time

size = 1000000

python_list = list(range(size))
numpy_array = np.arange(size)
```

```python
start = time.time()
result = [x * 2 for x in python_list]
list_time = time.time() - start
```

```python
start = time.time()
result = numpy_array * 2
numpy_time = time.time() - start
```

```python
print('Time taken by List: ', list_time)
print('Time taken by Numpy: ', numpy_time)
```
